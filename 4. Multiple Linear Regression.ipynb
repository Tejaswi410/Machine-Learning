# Custom Multiple linear regression function

```python
class Multiple_Linear_Regression:
    
    def __init__(self):
        self.intercept_ = None
        self.coef_ = None 

    def fit(self,x_train,y_train):
        x_train = np.insert(x_train,0, 1, axis = 1)  #for introducing intercept term in the data


        #beta = ((inverse of ((transpose of x)*x)) * (transpose of x)*(y))
        beta = np.linalg.inv(np.dot(x_train.T,x_train)).dot(x_train.T).dot(y_train) #formula for finding beta values(first value will be intercept and rest are the coefficient)
        self.intercept_ = beta[0]
        self.coef_ = beta[1:]

    def predict(self,x_test):
        y_pred = np.dot(x_test,self.coef_) + self.intercept_
        return y_pred
```

```python
mlr = Multiple_Linear_Regression()
```

```python
x_train = np.array(x_train)
y_train = np.array(y_train)
x_test = np.array(x_test)
y_test = np.array(y_test)
```

```python
mlr.fit(x_train,y_train)
```

```python
y_pred = mlr.predict(x_test)
```

```python
r2_score(y_test,y_pred)
```
